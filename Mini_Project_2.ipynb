```python
predata['Dt_Customer']=pd.to_datetime(predata['Dt_Customer'],format="%Y-%m-%d")
```

```python
predata.info()
```

```python
#Calculating number of days from current_date

predata["Customer_Tenure"]=(
    pd.Timestamp.today() - predata["Dt_Customer"]
).dt.days
```

```python
predata.head()
```

```python
#Age from Year_Birth.

predata["age"]= pd.Timestamp.today().year - predata["Year_Birth"]

#Total_Spend as sum of all Mnt* fields.

predata["Total_Spend"]=predata['MntWines']+predata['MntFruits']+predata['MntMeatProducts']+predata['MntFishProducts']+predata['MntSweetProducts']+predata['MntGoldProds']

#Total_Purchases as sum of channel purchases.

predata['Total_Purchases']=predata['NumDealsPurchases']+predata['NumWebPurchases']+predata['NumCatalogPurchases']+predata['NumStorePurchases']

#Children = Kidhome + Teenhome.

predata['Children']=predata['Kidhome']+predata['Teenhome']
```
