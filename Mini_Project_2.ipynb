```python
# Segmentation task

predata1.loc[:,"High_Income"]=(predata1["Income"]>75000).astype(int)

predata1.loc[:,"Young_Customers"]=(predata1["age"]<30).astype(int)

predata1.loc[:,"Campaign_Responders"]=(predata1["Response"]==1).astype(int)

predata1.loc[:,"High_Web_Engagement"]=(predata1["NumWebVisitsMonth"]>5).astype(int)

predata1.loc[:,"Family_Customer"]=(predata1["Children"]>0).astype(int)

threshold = predata1["Total_Spend"].quantile(0.90)
predata1["High_Spender"] = (predata1["Total_Spend"] > threshold).astype(int)
```

```python
''' So after segmentation, I calculated % of customers belonging to each segment.
I got an interpretation 0% of customers are young, 35.3% are high income,10% are top spenders,53.5% are highly engaged,
68.3% have children,14.7% accepted campaign. From the interpretation, I need to implement a segment. To choose the segment,
I calculated response rate for each segment. Segments were evaluated using lift relative to the overall response rate.
Only attributes with lift greater than 1 were included in the Ideal Target definition to maximize campaign response and ROI.

#Lift Interpretation table for reference

Lift value	    Meaning	                    Action
> 1.2	          Strong positive signal	    Include
1.0 – 1.2	      Weak positive	              Optional
≈ 1.0	          No effect	                  Exclude
< 1.0	          Negative	                  Exclude
```

```python
overall_response = predata1["Response"].mean()
print(overall_response)
```

```python
segments = [
    "High_Income",
    "High_Spender",
    "High_Web_Engagement",
    "Family_Customer",
    "Campaign_Responders"
]

for seg in segments:
    print(predata1.groupby(seg)["Response"].mean())
```

```python
for seg in segments:
    resp_rate = predata1.groupby(seg)["Response"].mean()
    lift = resp_rate / overall_response
    print(f"\n{seg} lift:")
    print(lift)
```

```python
# Master segment
predata1["Ideal_Target"] = (
    (predata1["High_Income"] == 1) &
    (predata1["High_Spender"] == 1) &
    (predata1["Campaign_Responders"] == 1)
).astype(int)
```
